```python
#Task 3: Exploratory Data Analysis (EDA)
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Users\DELL\Desktop\netflix_titles.csv")
print("First 5 Rows:")
print(df.head())
```

```python
# TASK 1 : SUMMARY STATISTICS
print("\nSummary Statistics:")
print(df.describe())
```

```python
# TASK 2 : CORRELATION ANALYSIS
df["release_year"] = pd.to_numeric(df["release_year"])
numeric_df = df[["release_year"]]
correlation = numeric_df.corr()
print("\nCorrelation Matrix:")
print(correlation)
```

```python
# TASK 3 : TREND IDENTIFICATION
#Trend 1: Movies vs TV shows
type_count = df["type"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(type_count.index, type_count.values)
plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

```python
# TREND 2
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
df["year_added"] = df["date_added"].dt.year
yearly_content = df["year_added"].value_counts().sort_index()
plt.figure(figsize=(8,4))
plt.plot(yearly_content.index, yearly_content.values)
plt.title("Content Added to Netflix Over Time")
plt.xlabel("Year")
plt.ylabel("Number of Titles Added")
plt.show()
```

```python
# TREND 3
country_count = df["country"].dropna().value_counts().head(10)
plt.figure(figsize=(8,5))
plt.bar(country_count.index, country_count.values)
plt.title("Top 10 Content Producing Countries")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
print("\nEDA Completed Successfully!")
```
